```python
import torch
from torch import nn
from mmcv import Timer
```

```python
input_size = (224, 224, 224)
in_channels = 32
out_channels = 32
stride = 1
kernel = 5

conv_layer = nn.Conv3d(in_channels, out_channels, 
                        kernel_size=kernel, 
                        padding=(kernel - 1)//2, 
                        stride = stride)

dwconv_layer = nn.Sequential(
                    nn.Conv3d(in_channels, out_channels, 
                            kernel_size=kernel, 
                            padding=(kernel - 1)//2, 
                            stride = stride, 
                            groups=in_channels), 
                    nn.Conv3d(out_channels, out_channels, 
                            kernel_size=1)
                        )

```

```python


input_tensor = torch.rand((1, in_channels) + input_size, device='cuda:5', dtype = torch.float)
conv_layer = conv_layer.cuda(device='cuda:5')
dwconv_layer = dwconv_layer.cuda(device = 'cuda:5')
# conv_layer.eval()
# dwconv_layer.eval()

```

```python
timer = Timer()
timer.start()
num_run = 20
for i in range(num_run):
    conv_res = conv_layer(input_tensor)

duration = timer.since_last_check()
print(f'Conv 100 total {duration} per {duration /num_run}')
```

```python
timer = Timer()
timer.start()
for i in range(num_run):
    dwconv_res = dwconv_layer(input_tensor)
duration = timer.since_last_check()
print(f'DWConv 100 total {duration} per {duration /num_run}')
```
